```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

# Given Datasets
train_x = [12, 14, 17, 20, 23, 26, 30, 34, 37, 39]
train_y = [5, 10, 23, 27, 36, 39, 45, 42, 44, 40]
test_x = [13, 21, 32, 38]
test_y = [7, 32, 43, 43]
degrees = [1, 2]

def evaluate_degrees(train_x, train_y, test_x, test_y, degrees):
    # Step 1: Reshape 1D input lists into 2D column matrices (-1 rows, 1 column)
    train_x = np.array(train_x, dtype=float).reshape(-1, 1)
    train_y = np.array(train_y, dtype=float)
    test_x = np.array(test_x, dtype=float).reshape(-1, 1)
    test_y = np.array(test_y, dtype=float)

    results = []

    # Step 2: Loop through each polynomial degree
    for degree in degrees:
        # Build pipeline and train on (train_x, train_y)
        pipeline = make_pipeline(PolynomialFeatures(degree), LinearRegression())
        pipeline.fit(train_x, train_y)

        # Predict sales for both training set and test set
        train_pred = pipeline.predict(train_x)
        test_pred = pipeline.predict(test_x)

        # Calculate errors (MSE) and round to 2 decimals
        train_mse = round(float(mean_squared_error(train_y, train_pred)), 2)
        test_mse = round(float(mean_squared_error(test_y, test_pred)), 2)

        # Calculate absolute gap between test error and train error
        gap = round(abs(test_mse - train_mse), 2)

        # Save metrics for this degree
        results.append({
            "degree": degree,
            "train_mse": train_mse,
            "test_mse": test_mse,
            "gap": gap
        })

    # Step 3: Find the degree with the smallest generalization gap
    best_degree = min(results, key=lambda res: res["gap"])["degree"] if results else None

    return {"results": results, "best_degree": best_degree}

if __name__ == "__main__":
    print(evaluate_degrees(train_x, train_y, test_x, test_y, degrees))
```
